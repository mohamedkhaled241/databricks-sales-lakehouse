#Init

```python
import pyspark.sql.functions as F
from pyspark.sql.window import Window
```

#Reading From Silver Tables

```python

df = spark.table("workspace.silver.crm_customers")
```

# Create a standardized customer key for joining location with customer

```python
df = df.withColumn(
    "location_customer_key",
    F.regexp_replace("customer_key", r"^(.{2})", "$1-")
)
```

```python
df.show()
```

#Write it to Gold Table

```python
(
    df.write
       .mode("overwrite")
       .format("delta")
       .option("overwriteSchema", "true")
       .saveAsTable("gold.dim_customers")

)
```

```python
%sql
select * from gold.dim_customers
```
